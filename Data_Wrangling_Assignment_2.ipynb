```python
def inspect_numeric_table(file_path, sep="\t"):
    print("\nReading file:", file_path.resolve())

    raw = pd.read_csv(
        first_file,
        sep=sep,
        header=0,
        dtype=str,
        encoding="utf-8-sig",
        na_values=["", "NA", "NaN", "nan", "null", "NULL"],
        keep_default_na=True,
        on_bad_lines="warn",
    )

    numeric = raw.apply(pd.to_numeric, errors="coerce")

    numeric_columns = [
        col for col in raw.columns
        if numeric[col].notna().any()
    ]

    print("Shape:", raw.shape)
    print("Columns:", raw.columns.tolist())
    print("Numeric candidate columns:", numeric_columns)
    print("\nFirst 10 rows:")
    print(raw.head(10).to_string(index=False))

    return raw, numeric

raw, numeric = inspect_numeric_table(first_file)
```

```python
def knn_impute(X, k=15):
  X = np.asarray(X, dtype=float)

  if X.ndim != 2:
    raise ValueError("X must be a 2D matrix.")
  if k < 1:
    raise ValueError("k must be at least 1.")
  if np.isinf(X).any():
    raise ValueError("X must not contain infinite values.")

  result = X.copy()
  n_genes, n_experiments = X.shape

  all_observed = X[np.isfinite(X)]
  if all_observed.size == 0:
    raise ValueError("X contains no observed values.")

  global_mean = all_observed.mean()

  for i in range(n_genes):
    missing_columns = np.flatnonzero(np.isnan(X[i]))

    for c in missing_columns:
      distances = []

      for j in range(n_genes):
        if i == j or np.isnan(X[j,c]):
          continue

        shared = (
            np.isfinite(X[i]) &
            np.isfinite(X[j])
        )

        n_shared = shared.sum()
        if n_shared == 0:
          continue

        diff = X[i, shared] - X[j, shared]

        distance = np.sqrt(
            np.sum(diff ** 2) *
            n_experiments / n_shared
        )

        distances.append((distance, j))


      if not distances:
        observed = X[i, np.isfinite(X[i])]
        result[i, c] = (
            observed.mean()
            if observed.size
            else global_mean
        )
        continue

      distances.sort(key=lambda pair: pair[0])
      neighbors = distances[:k]

      exact = [
          j for d, j in neighbors if d < 1e-12
      ]

      if exact:
        result[i, c] = np.mean(X[exact, c])
        continue

      weights = np.array([
          1.0 / d for d, j in neighbors
      ])

      values = np.array([
          X[j, c] for d, j in neighbors
      ])

      result[i, c] = np.average(
          values, weights=weights
      )

  return result
```

```python
def row_average_impute(X):
  X = np.asarray(X, dtype=float).copy()

  if X.ndim != 2:
    raise ValueError("X must be a 2D matrix.")

  X[~np.isfinite(X)] = np.nan

  result = X.copy()
  observed_mask = np.isfinite(X)
  all_observed = X[observed_mask]

  if all_observed.size == 0:
    raise ValueError("X contains no observed values.")

  global_mean = all_observed.mean()

  row_counts = observed_mask.sum(axis=1)
  row_sums = np.where(observed_mask, X, 0.0).sum(axis=1)

  row_means = np.full(X.shape[0], global_mean)
  valid_rows = row_counts > 0

  row_means[valid_rows] = (
      row_sums[valid_rows] / row_counts[valid_rows]
  )

  missing_rows, missing_cols = np.where(~observed_mask)
  result[missing_rows, missing_cols] = row_means[missing_rows]

  if not np.isfinite(result).all():
    raise ValueError("imputation did not produce finite values.")

  return result
```

```python
def zero_impute(X):
  X = np.asarray(X, dtype=float)

  if X.ndim != 2:
    raise ValueError("X must be a 2D matrix.")

  if np.isinf(X).any():
    raise ValueError("X contains infinite values.")

  result = X.copy()
  result[np.isnan(result)] = 0.0

  return result
```

```python
def mask_random_values(X, missing_fraction, seed=42):
  X = np.asarray(X, dtype=float)

  if X.ndim != 2:
    raise ValueError("X must be a 2D matrix.")

  if not 0 <= missing_fraction <= 1:
    raise ValueError(
        "missing_fraction must be between 0 and 1."
    )

  if np.isinf(X).any():
    raise ValueError("X must not contain infinite values.")

  rng = np.random.default_rng(seed)
  masked = X.copy()

  observed_indices = np.flatnonzero(np.isfinite(X))

  n_missing = int(len(observed_indices) * missing_fraction)

  selected = rng.choice(
      observed_indices,
      size=n_missing,
      replace=False
  )

  mask = np.zeros(X.shape, dtype=bool)
  mask.flat[selected] = True

  masked.flat[selected] = np.nan

  return masked, mask
```

```python
def normalized_rms_error(original, imputed, mask):
  original = np.asarray(original, dtype=float)
  imputed = np.asarray(imputed, dtype=float)
  mask = np.asarray(mask, dtype=bool)

  if original.shape != imputed.shape or original.shape != mask.shape:
    raise ValueError("All arrays must have the same shape.")

  true_values = original[mask]
  predicted_values = imputed[mask]

  if true_values.size == 0:
    raise ValueError("No masked values to evaluate.")

  if not np.isfinite(true_values).all():
    raise ValueError("Masked reference values must be finite.")

  if not np.isfinite(predicted_values).all():
    raise ValueError("Some masked values are still unimputed")

  rms_error = np.sqrt(
      np.mean((true_values - predicted_values) ** 2)
  )

  observed_values = original[np.isfinite(original)]

  if observed_values.size == 0:
    raise ValueError("Orignial matrix contains no observed values.")

  reference_rms = np.sqrt(
      np.mean(observed_values ** 2)
  )

  if reference_rms < 1e-12:
    raise ValueError("Reference RMS is too small.")

  return rms_error / reference_rms
```

```python
import matplotlib.pyplot as plt

X_original = np.asarray(X, dtype=float).copy()
X_original[~np.isfinite(X_original)] = np.nan

print("Genes:", X.shape[0])
print("Experiments:", X.shape[1])
print("Original missing fraction:", np.isnan(X).mean())
print("Finite measurements:", np.isfinite(X_original).sum())

if not np.isfinite(X_original).any():
  raise ValueError("The matrix contains no finite measurements.")

missing_fractions = [0.01, 0.05, 0.10, 0.15, 0.20]
seeds = range(10)

methods = {
    "KNN (k=15)": lambda M: knn_impute(M, k=15),
    "row average": row_average_impute,
    "zero fill": zero_impute,
}

results = []

for fraction in missing_fractions:
  for seed in seeds:
    masked, mask = mask_random_values(
        X_original,
        missing_fraction=fraction,
        seed=seed
    )

    for method_name, imputer in methods.items():
      estimate = imputer(masked)

      if not np.isfinite(estimate).all():
        raise ValueError(
            f"{method_name} produced non-finite values"
            f"at fraction= {fraction}, seed={seed}."
        )

      error = normalized_rms_error(
          X_original,
          estimate,
          mask
      )

      results.append({
          "dataset": "Spellman",
          "missing_fraction": fraction,
          "seed": seed,
          "method": method_name,
          "NRMS": error,
      })

results_df = pd.DataFrame(results)

summary = (
    results_df
    .groupby(["dataset", "missing_fraction", "method"])
    ["NRMS"]
    .agg(["mean", "std"])
    .reset_index()
)

print("\nSummary of experimental results:")
print(summary.to_string(index=False))

results_df.to_csv("Spellman_all_runs.csv", index=False)
summary.to_csv("spellman_results.csv", index=False)
```

```python
for dataset_name in summary["dataset"].unique():
  subset_dataset = summary[
      summary["dataset"] == dataset_name
  ]

  fig, ax = plt.subplots(figsize=(8, 5))

  for method_name in methods:
    subset = subset_dataset[
        subset_dataset["method"] == method_name
    ].sort_values("missing_fraction")

    ax.errorbar(
        subset["missing_fraction"] * 100,
        subset["mean"],
        yerr=subset["std"].fillna(0),
        marker="o",
        capsize=3,
        label=method_name,
    )

ax.set_xlabel("Artificially missing vlaues (%)")
ax.set_ylabel("Normalized RMS error")
ax.set_title(f"Imputation performance: {dataset_name}")
ax.legend()
fig.tight_layout()
fig.savefig(
    f"{dataset_name.lower()}_results.png",
    dpi=200
    )

plt.show()
plt.close(fig)
```
